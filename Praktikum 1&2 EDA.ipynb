```python
# Import library
# Hadijah Tohepaly_202431078

import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Membaca dan menampilkan dataset
# Hadijah Tohepaly_202431078

df = pd.read_csv("housing.csv")

df
```

```python
# Melihat deskripsi statistik
# Hadijah Tohepaly_202431078

df.describe()
```

```python
# Menghitung missing value
# Hadijah Tohepaly_202431078

df.isna().sum()
```

```python
# Melihat nilai unik kolom kategorik
# Hadijah Tohepaly_202431078

df["ocean_proximity"].unique()
```

```python
# Menangani missing value
# Hadijah Tohepaly_202431078

# Mengisi missing value kolom numerik dengan mean
for kolom in df.select_dtypes(include="number").columns:
    df[kolom] = df[kolom].fillna(df[kolom].mean())

# Mengisi missing value kolom kategorik dengan modus
for kolom in df.select_dtypes(include="object").columns:
    df[kolom] = df[kolom].fillna(df[kolom].mode()[0])
```

```python
df.isna().sum()
```

```python
# Menghitung data duplikat
# Hadijah Tohepaly_202431078

df.duplicated().sum()
```

```python
# Menghapus data duplikat
# Hadijah Tohepaly_202431078
df = df.drop_duplicates()

df.info()
```

```python
df.duplicated().sum()
```

```python
# Visualisasi data menggunakan diagram batang
# Hadijah Tohepaly_202431078

df["ocean_proximity"].value_counts().plot(kind="bar")

plt.title("Jumlah Data Berdasarkan Ocean Proximity")
plt.xlabel("Ocean Proximity")
plt.ylabel("Jumlah Data")
plt.xticks(rotation=45)
plt.show()
```
