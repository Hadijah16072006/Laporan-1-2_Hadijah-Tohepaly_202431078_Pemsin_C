```python
# Import library
# Hadijah Tohepaly_202431078

import requests
import csv
from bs4 import BeautifulSoup
import pandas as pd
```

```python
# Mengambil halaman website
# Hadijah Tohepaly_202431078

url = "https://magicscraper.app/demo/bulk"

headers = {
    "User-Agent": "Mozilla/5.0"
}

response = requests.get(url, headers=headers)

print("Status:", response.status_code)
```

```python
# Mengubah HTML menjadi objek BeautifulSoup
# Hadijah Tohepaly_202431078

soup = BeautifulSoup(response.text, "html.parser")

print(soup.prettify()[:3000])
```

```python
# Mencari tabel produk
# Hadijah Tohepaly_202431078

table = soup.find("table")

print("Tabel ditemukan:", table is not None)

rows = table.find_all("tr")

print("Jumlah baris:", len(rows))
```

```python
# Scraping 200 data produk
# Hadijah Tohepaly_202431078

result = []

for halaman in range(1, 5):

    if halaman == 1:
        url = "https://magicscraper.app/demo/bulk"
    else:
        url = f"https://magicscraper.app/demo/bulk{halaman}"

    response = requests.get(url, headers=headers)

    soup = BeautifulSoup(response.text, "html.parser")

    table = soup.find("table")

    rows = table.find_all("tr")

    print(f"Halaman {halaman}: {len(rows)-1} baris")

    for row in rows[1:]:

        cols = row.find_all("td")

        if len(cols) >= 11:

            result.append({
                "sku": cols[0].get_text(strip=True),
                "nama_produk": cols[1].get_text(strip=True),
                "maker": cols[2].get_text(strip=True),
                "kategori": cols[3].get_text(strip=True),
                "pack": cols[4].get_text(strip=True),
                "harga": cols[5].get_text(strip=True),
                "harga_sebelumnya": cols[6].get_text(strip=True),
                "stock": cols[7].get_text(strip=True),
                "lead_time": cols[8].get_text(strip=True),
                "rating": cols[9].get_text(strip=True),
                "reviews": cols[10].get_text(strip=True)
            })

print("Jumlah data:", len(result))
```

```python
# Menampilkan hasil scraping
# Hadijah Tohepaly_202431078

for data in result[:10]:
    print(data)
```

```python
# Mencari link produk
# Hadijah Tohepaly_202431078

for tag in soup.find_all("a", href=True):
    print(tag.get_text(strip=True), tag["href"])
```

```python
# Menyimpan hasil scraping ke CSV
# Hadijah Tohepaly_202431078

with open("hasil_scraping.csv", "w",
          newline="", encoding="utf-8") as file:

    fieldnames = [
        "sku", "nama_produk", "maker", "kategori",
        "pack", "harga", "harga_sebelumnya",
        "stock", "lead_time", "rating", "reviews"
    ]

    writer = csv.DictWriter(
        file, fieldnames=fieldnames
    )

    writer.writeheader()
    writer.writerows(result)

print("CSV berhasil disimpan")
```

```python
# Membaca hasil scraping
# Hadijah Tohepaly_202431078

df = pd.read_csv("hasil_scraping.csv")

df.head()
```

```python
# Mengubah kolom numerik
# Hadijah Tohepaly_202431078

df["harga"] = pd.to_numeric(
    df["harga"].str.replace(r"[^0-9.]", "", regex=True),
    errors="coerce"
)

df["rating"] = pd.to_numeric(
    df["rating"].str.extract(r"(\d+\.\d+)")[0],
    errors="coerce"
)

df["reviews"] = pd.to_numeric(
    df["reviews"], errors="coerce"
)

df["pack"] = pd.to_numeric(
    df["pack"], errors="coerce"
)

df.info()
```

```python
# Menyimpan dataset akhir
# Hadijah Tohepaly_202431078

df.to_csv("hasil_scraping.csv", index=False)

print("Jumlah data:", len(df))

df.head()
```

```python
# Membersihkan kolom harga sebelumnya
# Hadijah Tohepaly_202431078

df["harga_sebelumnya"] = pd.to_numeric(
    df["harga_sebelumnya"].astype(str).str.replace(
        r"[^0-9.]", "", regex=True
    ),
    errors="coerce"
)

df.info()
```

```python
# Menyimpan dataset final
# Hadijah Tohepaly_202431078

df.to_csv("hasil_scraping.csv", index=False)

print("Jumlah data:", len(df))
print("Dataset berhasil disimpan")
```
